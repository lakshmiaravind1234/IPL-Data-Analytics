## Data Cleaning

```python
exclude_cols = ['method', 'city']
```

```python
cat_cols = matches.select_dtypes(include='object').columns.difference(exclude_cols)
```

Observation:
* Here cat_cols are separated but we excluded two columns those are method col,city col because those are very important columns.in city column if we fill with mode value then different venue may contain different city .
* **Example:**
* Venue:Dharmasala  City:Himachal Pradesh
* Venue:Dharmasala  City:Himachal Pradesh
* Venue:Chinnaswamy City:Nan
* Venue:Chinnaswamy City:Banglore
 * When we fill with mode the most repeated city is Himachal Pradesh
the city would filled with Himachal Pradesh but actually the Chinnaswamy Stadium is in the Banglore City. Thats why we exclude the city column.

```python
for col in cat_cols:
    matches[col] = matches[col].fillna(matches[col].mode()[0])
```

**Outliers**

```python
num_cols = matches.select_dtypes(include='number').columns.drop('id')
```

```python
num_cols
```

```python

for col in num_cols:
    Q1 = matches[col].quantile(0.25)
    Q3 = matches[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    median = matches[col].median()

    matches.loc[
        (matches[col] < lower_bound) | (matches[col] > upper_bound),
        col
    ] = median
```

Observation : The null values in columns such as player_dismissed, dismissal_kind, fielder, and extras_type were not filled because the missing values have a meaningful representation in the IPL delivery-level data.
* For example, a null value in player_dismissed means that no wicket occurred on that delivery.
* while a null value in dismissal_kind means that no dismissal took place.
* Similarly, a null value in fielder indicates that no fielder was involved or recorded
* A null value in extras_type indicates that no extra was recorded.
* Filling these null values with the mode or an arbitrary value could introduce incorrect information and distort the analysis. Therefore, the meaningful null values were retained.
